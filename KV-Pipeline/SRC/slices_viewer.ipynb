```python
import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import interact, IntSlider, Dropdown

# Load volume
raw_vol = main_load_volume(vol_path)

vol = process_volume_base(
    raw_vol,
    align_orientation=align_orient,
    max_clip=max_clip
)

print(vol.shape)
```

```python
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider, Dropdown

def show_slice(axis, index):

    fig, ax = plt.subplots(figsize=(6,6))

    if axis == "X":
        img = vol[index,:,:]

    elif axis == "Y":
        img = vol[:,index,:]

    else:
        img = vol[:,:,index]

    ax.imshow(
        img,
        cmap="gray",
        vmin=0,
        vmax=max_clip
    )

    ax.set_title(f"{axis} Slice {index}")
    ax.axis("off")

    plt.show()


interact(
    show_slice,

    axis=Dropdown(
        options=["X","Y","Z"],
        value="Z"
    ),

    index=IntSlider(
        min=0,
        max=vol.shape[2]-1,
        step=1,
        value=vol.shape[2]//2
    )
);
```

```python
def show_every_n(axis="Z", step=8):

    if axis=="X":
        total = vol.shape[0]

    elif axis=="Y":
        total = vol.shape[1]

    else:
        total = vol.shape[2]

    ids = list(range(0,total,step))

    cols = 4
    rows = int(np.ceil(len(ids)/cols))

    fig,axs = plt.subplots(
        rows,
        cols,
        figsize=(4*cols,4*rows)
    )

    axs=np.array(axs).reshape(-1)

    for ax in axs:
        ax.axis("off")

    for ax,i in zip(axs,ids):

        if axis=="X":
            img=vol[i,:,:]

        elif axis=="Y":
            img=vol[:,i,:]

        else:
            img=vol[:,:,i]

        ax.imshow(
            img,
            cmap="gray",
            vmin=0,
            vmax=max_clip
        )

        ax.set_title(f"{axis}={i}")
        ax.axis("off")

    plt.tight_layout()
    plt.show()


interact(
    show_every_n,

    axis=Dropdown(
        options=["X","Y","Z"],
        value="Z"
    ),

    step=IntSlider(
        min=1,
        max=32,
        value=8
    )
);
```

```python
from ipywidgets import Play, IntSlider, jslink, HBox, VBox, interactive_output

play = Play(
    value=0,
    min=0,
    max=vol.shape[2]-1,
    step=1,
    interval=150
)

slider = IntSlider(
    min=0,
    max=vol.shape[2]-1
)

jslink((play,'value'),(slider,'value'))

def update(i):

    plt.figure(figsize=(6,6))
    plt.imshow(
        vol[:,:,i],
        cmap="gray",
        vmin=0,
        vmax=max_clip
    )
    plt.title(f"Slice {i}")
    plt.axis("off")
    plt.show()

out = interactive_output(update, {'i': slider})

VBox([HBox([play,slider]), out])
```
